#### **Situação 8:**

Recebemos a demanda de criar uma lista com a média dos estudantes da lista de listas que criamos na Situação 6, arredondando a média para uma casa decimal. Lembrando que cada lista da lista de listas contém as três notas de cada estudante.

**Vamos resolver esse desafio?**

**Tip:** Utiliza el formato:
```python
[expressão for item in lista]
```

````python
def média(lista: list=[0]) -> float:
  ''' Função para calcular a média de notas en una lista

  lista: list, default [0]
    Lista con las notas para calcular a média
  return = calculo: float
    Média calculado
  '''

  calculo = sum(lista) / len(lista)

  return calculo
````

````python
notas = [[8.0, 9.0, 10.0], [9.0, 7.0, 6.0], [3.4, 7.0, 7.0], [5.5, 6.6, 8.0], [6.0, 10.0, 9.5]]
````

````python
médias = [round(média(nota),1) for nota in notas]
médias
````

## 4.1 Tratando as Exceções

O tratamento de exceções contribui para estabelecer um fluxo alternativo para a execução do código, evitando a interrupção dos processos de forma inesperada.

Existe uma série de exceções, e a partir do comportamento que queremos e dos erros que queremos tratar, é possível construir um caminho para o usuário ou também fornecer mais detalhes sobre aquela exceção.

- Hierarquia das exceções (https://docs.python.org/pt/3/library/exceptions.html#exception-hierarchy)

### Try ... Except

```python
try:
  # código que será executado. Caso surja uma exceção, pare imediatamente
except <nombre_de_la_excepcion as e>:
  # Se surgir uma exceção no try, execute este código; caso contrário, pule esta etapa
```

Você criou um código que lê um dicionário com as notas dos estudantes e queria retornar a lista de notas de um estudante.

**Vamos testar este primeiro tratamento?**



````python
notas = {'João': [8.0, 9.0, 10.0], 'Maria': [9.0, 7.0, 6.0], 'José': [3.4, 7.0, 8.0], 'Claudia': [5.5, 6.6, 8.0],
         'Ana': [6.0, 10.0, 9.5], 'Jorge': [5.5, 7.5, 9.0], 'Julia': [6.0, 8.0, 7.0], 'Pedro': [3.0, 4.0, 6.0]}
````

````python
nome = input('Digite o nome do/a estudante: ')
resultado = notas[nome]
resultado
````

#### **Situação 15:**

Você criou uma função para calcular a média de um estudante em uma determinada disciplina, passando em uma lista as notas desse estudante.

Pretendes tratar 2 situaçãoes:

- Se a lista possuir um valor não numérico, o cálculo da média não será executado e uma mensagem de "Não foi possível calcular a média do(a) estudante. Só são permitidos valores numéricos!" será exibida.

- Se a lista tiver mais de 4 notas, surgirá uma exceção do tipo **ValueError** informando que "A lista não pode ter mais de 4 notas."

Un texto avisando que "A consulta foi concluída." debe ser mostrado independientemente de si surgió o no alguna excepción.

**Vamos resolver esse desafio?**

````python
def média(lista: list=[0]) -> float:
  ''' Função para calcular a média de notas en una lista

  lista: list, default [0]
    Lista con las notas para calcular a média
  return = calculo: float
    Média calculado
  '''

  calculo = sum(lista) / len(lista)

  if len(lista) > 4:
    raise ValueError('A lista não pode ter mais de 4 notas.')

  return calculo
````

````python
notas = [6,7,8,9,10]
resultado =  média(notas)
resultado
````

````python
notas = [6,7,8,9,'10']
resultado =  média(notas)
resultado
````
