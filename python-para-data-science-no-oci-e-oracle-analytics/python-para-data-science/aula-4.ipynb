#Ciclos e laços

```
#Laço fora da função
idadees = [18,15,12,45,50]

def verificar_se_pode_dirigir(idade):
  if idade >= 18:
    print("Você tem idade suficiente para dirigir")
  else:
    print("Você NÃO tem idade suficiente para dirigir")


for idade in idadees:
  verificar_si_puede_conducir(idade)
```

```
#Bloco dentro da função
idadees = [18,15,12,45,50]

def verificar_si_puede_conducir_con_bucle(idadees):
  for idade in idadees:
    if idade >= 18:
      print("Você tem idade suficiente para dirigir")
    else:
      print("Você NÃO tem idade suficiente para dirigir")

verificar_si_puede_conducir_con_bucle(idadees)
```

```
verificaciones = []
idadees = [13,15,20]

def verificar_si_puede_conducir_bool(verificaciones,idadees):
  for idade in idadees:
    if idade >= 18:
      verificaciones.append(True)
    else:
      verificaciones.append(False)
  for verificacion in verificaciones:
    if verificacion == True:
      print('Você tem idade suficiente para dirigir')
    else:
      print("Você NÃO tem idade suficiente para dirigir")

verificar_si_puede_conducir_bool(verificaciones,idadees)
```
