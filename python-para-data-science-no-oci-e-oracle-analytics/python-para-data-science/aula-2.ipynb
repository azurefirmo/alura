#Condicionais

```
idade = 18

def verificar_se_pode_dirigir(idade):
  if idade >= 18:
    print("Você tem idade suficiente para dirigir")
  else:
    print("Você NÃO tem idade suficiente para dirigir")
  
verificar_si_puede_conducir(idade)
```

#Condicionais sem parâmetro e conversão de tipo de dado

```
def verificar_se_pode_dirigir_sem_parametro():
  idade = input('Digite sua idade: ')
  idade = int(idade)
  if idade >= 18:
    print("Você tem idade suficiente para dirigir")
  else:
    print("Você NÃO tem idade suficiente para dirigir")

verificar_si_puede_conducir_sin_parametro()
```
